```python
import pickle
import numpy as np
from collections import defaultdict

with open("/kaggle/working/cache/communities/leiden_train.pkl", "rb") as f:
    leiden_membership = pickle.load(f)

print(type(leiden_membership))
print(f"Accounts: {len(leiden_membership):,}")

# Invert to community -> members
groups = defaultdict(list)
for account, community in leiden_membership.items():
    groups[community].append(account)

sizes = np.array([len(members) for members in groups.values()])

print(f"Communities: {len(sizes):,}")
print(f"Largest: {sizes.max():,}")
print(f"Smallest: {sizes.min():,}")
print(f"Mean: {sizes.mean():.2f}")
print(f"Median: {np.median(sizes):.0f}")
print(f"95th percentile: {np.percentile(sizes,95):.0f}")
print(f"99th percentile: {np.percentile(sizes,99):.0f}")

largest = sorted(sizes, reverse=True)[:20]
print("\nTop 20 largest communities:")
print(largest)

print("\nCommunities larger than:")
for threshold in [100, 500, 1000, 5000, 10000]:
    print(f">{threshold:5d}: {(sizes > threshold).sum()}")
```

```python
import gc, os, psutil, resource
_proc = psutil.Process(os.getpid())
def live_gb(): return _proc.memory_info().rss / 1024**3
def peak_gb(): return resource.getrusage(resource.RUSAGE_SELF).ru_maxrss / 1024**2 / 1024
def mem(tag): print(f"[{tag}] live={live_gb():.1f} GB  peak={peak_gb():.1f} GB")

mem("start of holdout cell")

# ---- 1. FREE everything training left behind, BEFORE loading LI ----
# With build_features_from_holdout=True, score_holdout ignores df_train entirely
# (it builds node features from LI itself). df_train is the FULLY ENGINEERED HI
# frame -- ~5M rows x ~40 extra float cols + raw strings. Multiple GB, pure dead
# weight while we load LI (bigger than HI) and engineer ITS features on top.
keep_model, keep_threshold = result["model"], result["threshold"]
keep_feature_columns = result["feature_columns"]

del df_train                       # engineered HI frame -- not needed
del result                         # holds node_features (~0.5GB) + val_probs
for _v in ["src", "dst", "preds", "succs"]:   # transaction-graph CSRs from cell 5
    if _v in globals():
        del globals()[_v]
gc.collect()
mem("after freeing train artifacts")

# ---- 2. Load + engineer LI ----
df_test, _ = load_and_clean(str(TEST_CSV), vocabs=vocabs)
mem("after load_and_clean")

if COMBINE_WITH_BASE_FEATURES:
    _, src_t, dst_t, preds_t, succs_t = build_transaction_graph(df_test)
    df_test = engineer_all_features(df_test, preds_t, succs_t, src_t, dst_t)
    del src_t, dst_t, preds_t, succs_t     # CSR adjacency, dead after engineering
    gc.collect()
mem("after engineer_all_features")

# ---- 3. Score. df_train_plus_val=None is safe: ignored when
#         build_features_from_holdout=True (verified). ----
holdout = T.score_holdout(
    GRAPH_CACHE_DIR, COMMUNITY_CACHE_DIR, FEATURE_CACHE_DIR,
    keep_model, None, df_test, keep_feature_columns, keep_threshold,
    restrict_to_accounts=RESTRICT_TO_ACCOUNTS,
    build_features_from_holdout=True,
)
print(holdout["metrics"])
mem("after score_holdout")

# ---- 4. Exports (df_holdout_view is the slim 5-col frame) ----
del df_test
holdout.pop("node_features", None)     # LI account table, not needed for exports
gc.collect()

transaction_view = build_transaction_view(holdout["df_holdout_view"], holdout["probs"], threshold=keep_threshold)
account_view     = aggregate_accounts(holdout["df_holdout_view"], holdout["probs"], threshold=keep_threshold)
transaction_view.to_csv(EXPORT_DIR / "transaction_view.csv", index=False)
account_view.to_csv(EXPORT_DIR / "account_view.csv", index=False)   # <-- you were missing this
mem("done")
print("Wrote transaction_view.csv / account_view.csv")
```
